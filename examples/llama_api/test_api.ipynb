```python
from openai import AsyncOpenAI
```

```python
client = AsyncOpenAI(base_url="http://localhost:7776/v1", api_key="dummy-key")
l = await client.models.list()
```

```python
# Cell 3: Use chat completions directly
async def get_completion(prompt):
    response = await client.chat.completions.create(
        model="any-model-id",  # This is ignored by your server
        messages=[{"role": "user", "content": prompt}],
        stream=True,  # Important - your server requires streaming
        temperature=0.7
    )
    
    full_response = ""
    async for chunk in response:
        if chunk.choices and chunk.choices[0].delta.content:
            content = chunk.choices[0].delta.content
            full_response += content
            print(content, end="", flush=True)
    
    print("\n")
    return full_response

# Cell 4: Test it
prompt = "What is the capital of France? What's it's population"
response_text = await get_completion(prompt)
```
