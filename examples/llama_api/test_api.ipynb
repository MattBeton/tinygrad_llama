```python
# Cell 3: Use chat completions directly
async def get_completion(prompt):
    completion = await client.chat.completions.create(
        model="any-model-id",  # This is ignored by your server
        messages=[{"role": "user", "content": prompt}],
        tools=[
            {
                "type": "function",
                "function": {
                    "name": "get_capital",
                    "description": "Get the capital of a country",
                    "parameters": {
                    "properties": {
                        "country": {
                        "type": "string"
                        }
                    }
                }
                }
            }
        ],
        tool_choice="auto",
        max_tokens=1000,
        stream=False
    )

    assistant_message = completion.choices[0].message

    return assistant_message.content
    
# Cell 4: Test it
prompt = "What is the capital of France?"
response_text = await get_completion(prompt)
response_text
```

```python
from openai import OpenAI

def stream_chat_response(prompt):
    # Initialize client
    client = OpenAI(base_url="http://localhost:7776/v1", api_key="dummy-key")
    
    # Create a streaming completion
    stream = client.chat.completions.create(
        model="any-model-id",  # This is ignored by your local server
        messages=[{"role": "user", "content": prompt}],
        max_tokens=1000,
        stream=True  # Enable streaming
    )
    
    # Process and print the stream in real-time
    full_response = ""
    for chunk in stream:
        if chunk.choices and chunk.choices[0].delta.content is not None:
            content_chunk = chunk.choices[0].delta.content
            print(content_chunk, end="", flush=True)  # Print incrementally
            full_response += content_chunk
    
    return full_response

# Example usage - just run this in your cell
prompt = "Write a short poem about artificial intelligence."
stream_chat_response(prompt)
```
